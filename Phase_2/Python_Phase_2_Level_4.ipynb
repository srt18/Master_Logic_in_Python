###1. Print Four Stars (****)

```python
print("*"*4)
```

###2. Print n Stars on Same Line

```python
n = int(input("Enter the n stars: "))
print("*"*n)
```

###3. Print Square of Stars (n x n Stars)

```python
n = int(input("Enter the n stars: "))
for i in range(n):
  print("*"*n)
```

###4. Print an Increasing Triangle of Stars

```python
n = int(input("Enter the n stars: "))
for i in range(1,n+1):
  print("*"*i)
```

###5. Print a Right-Aligned Triangle of Stars

```python
n = int(input("Enter the n stars: "))
for i in range(1,n+1):
  print(" "*(n-i)+"*"*i)
```

###6. Print Stars in Even Numbers (2, 4, 6, 8, 10)

```python
n = int(input("Enter the n stars: "))
for i in range(1,n+1):
  print("*"*i*2)
```

###7. Print Stars in Odd Numbers (1, 3, 5, 7, 9)

```python
n = int(input("Enter the n stars: "))
for i in range(1,n+1):
  print("*"*(2*i-1))
```

##8. Print a Centered Pyramid of Stars

```python
n = int(input("Enter the n stars: "))
for i in range(1,n+1):
  print(" "*(n-i)+ "*"*(2*i-1) + " "*(n-i))
```

###9. Print Stars and Spaces Alternating (Stars and Blank Spaces)

```python
n = int(input("Enter the n stars: "))
for i in range(n):
  for j in range(n):
      if j < n -i -1:
        print("b", end="")
      else:
        if (j+i)%2 == 0:
          print("*", end="")
        else:
          print("b", end="")
        j+=1
  print()
```

###10. Print Numbers in an Increasing Sequence (1, 12, 123, 1234, 12345)

```python
n = int(input("N numbers pattern: "))

for i in range(1,n+1):
  for j in range(1,i+1):
    print(j,end="")
  print()
```

###11. Print Repeated Numbers per Row (Same Number Repeated)

```python
n = int(input("N numbers pattern: "))

for i in range(1,n+1):
  for j in range(1,i+1):
    print(i, end="")
  print()
```
